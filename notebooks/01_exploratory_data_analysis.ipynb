**Client:** Geldium Finance | **Project:** Delinquency Risk Prediction & AI Strategy

### Objectives:
1. Audit data quality and detect potential missingness or invalid credit distributions.
2. Profile borrower characteristics across income, credit limits, and debt ratios.
3. Analyze correlation structures between credit metrics and delinquency flags.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
# Load dataset
df = pd.read_csv('../data/delinquency_prediction_dataset.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

```python
# Data types and missing values summary
summary = pd.DataFrame({
    'DataType': df.dtypes,
    'NullCount': df.isnull().sum(),
    'NullPct': (df.isnull().sum() / len(df)) * 100
})
summary
```

```python
# Delinquency distribution
delinquency_rate = df['delinquent_flag'].mean() * 100
print(f'Overall Delinquency Rate: {delinquency_rate:.2f}%')

plt.figure(figsize=(6, 4))
sns.countplot(x='delinquent_flag', data=df, palette='Blues_r')
plt.title('Delinquency Flag Distribution (0: Performing, 1: Delinquent)')
plt.show()
```

```python
# Correlation matrix with Delinquent Flag
plt.figure(figsize=(10, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Credit Indicators')
plt.tight_layout()
plt.show()
```

### Key Insights from EDA:
- **Credit Utilization & Missed Payments:** Show strong positive correlation with default/delinquency risk.
- **Debt-to-Income Ratio (DTI):** Borrowers with DTI > 0.45 demonstrate higher vulnerability.
- **Data Integrity:** Dataset is complete with structured numeric features, ready for explainable risk modeling.
